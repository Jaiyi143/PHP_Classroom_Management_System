### 5.6 Outliers (IQR)

```python
def iqr_report(data, cols):
    rows=[]; masks={}
    for col in cols:
        s=data[col].dropna()
        q1=s.quantile(.25); q3=s.quantile(.75); iqr=q3-q1
        low=q1-1.5*iqr; high=q3+1.5*iqr
        mask=((data[col]<low)|(data[col]>high)).fillna(False)
        masks[col]=mask
        rows.append({"Column":col,"Q1":q1,"Q3":q3,"IQR":iqr,"Lower_Fence":low,
                     "Upper_Fence":high,"Outlier_Count":int(mask.sum())})
    return pd.DataFrame(rows), masks

outlier_report, outlier_masks = iqr_report(num, numeric_cols)
display(outlier_report)
for col in numeric_cols:
    if outlier_masks[col].any():
        print(col, "potential outliers:")
        display(df_raw.loc[outlier_masks[col], ["Record_ID", col]])
```

## 6. Data Cleaning and Transformation

The following cleaning strategy is applied (same as the Prelim and Midterm):

- Numeric fields are converted with `pd.to_numeric(errors="coerce")`.
- Numeric missing values are replaced with the **median**.
- Categorical missing values are replaced with `Unknown` (except `Typhoon_Impact`, where the missing values are the real label `None`).
- Known label variants are standardized with mapping dictionaries.
- Invalid dates are replaced with `Unknown`.
- Numeric outliers are replaced with the column **median** using the IQR rule.

```python
clean_df = df.copy()

categorical_columns = ["Record_ID", "Province", "Municipality", "Observation_Date", "Typhoon_Impact"]

# Convert numeric columns
for col in numeric_cols:
    clean_df[col] = pd.to_numeric(clean_df[col], errors="coerce")

# Fill numeric missing values with median
for col in numeric_cols:
    clean_df[col] = clean_df[col].fillna(clean_df[col].median())

# Typhoon_Impact: the "missing" values are the label None (no typhoon impact)
clean_df["Typhoon_Impact"] = clean_df["Typhoon_Impact"].fillna("None")

# Fill categorical missing values
for col in ["Province", "Municipality", "Observation_Date"]:
    clean_df[col] = clean_df[col].fillna("Unknown")

# Remove extra spaces
for col in categorical_columns:
    clean_df[col] = clean_df[col].str.strip()

# Standardize province
province_map = {
    "BENGUET": "Benguet",
    "Negros Occ.": "Negros Occidental"
}
clean_df["Province"] = clean_df["Province"].replace(province_map)

# Standardize municipality
municipality_map = {
    "SAN JOSE": "San Jose",
    "99999": "Unknown"
}
clean_df["Municipality"] = clean_df["Municipality"].replace(municipality_map)

# Invalid dates -> Unknown
valid_date = clean_df["Observation_Date"].str.match(date_pattern, na=False)
clean_df.loc[~valid_date, "Observation_Date"] = "Unknown"

# Handle numerical outliers using IQR + median
for col in numeric_cols:
    q1 = clean_df[col].quantile(0.25)
    q3 = clean_df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    median = clean_df[col].median()
    mask = (clean_df[col] < lower) | (clean_df[col] > upper)
    clean_df.loc[mask, col] = median

display(clean_df.head(10))
```

### Cleaning log

| Step | Column(s) | Action | Reason |

| 1 | Rainfall_mm, Temperature_C, Relative_Humidity_pct, Crop_Damage_ha | `pd.to_numeric(errors="coerce")` | placeholder text is not a number |

| 4 | Province, Municipality, Observation_Date | fill missing with `"Unknown"` | keep the record, do not guess |

| 7 | Municipality | `SAN JOSE`→`San Jose`, `99999`→`Unknown` | one spelling per municipality; code is not a place |
| 8 | Observation_Date | invalid dates → `Unknown` | a real date cannot be guessed |
| 9 | Numeric columns | IQR rule, outliers → median | removes error codes `999999.9` and `999.9` |

## 8. Data Quality Assessment (PASS / FAIL)

**Rules defined before evaluation**

| # | Quality dimension | Acceptance rule |
| --- | --- | --- |
| 1 | Completeness | no null values |
| 2 | Data type accuracy | numeric columns contain only numbers |
| 3 | Placeholder text | no `Unknown/NA` or `not_available` in numeric columns |
| 4 | Uniqueness | no duplicate `Record_ID` |
| 5 | Date format | every date is `Unknown` or a valid `2024-MM-DD` date |
| 6 | Consistency | Province, Municipality and Typhoon_Impact use only approved labels |
| 7 | Validity | rainfall 0–1000, temperature 10–45, humidity 0–100, crop damage ≥ 0 |
| 8 | Outlier review | no IQR outliers in numeric columns |

**Overall rule:** the overall result is PASS only if **all** rules pass; otherwise FAIL.

```python
def check_data(df):

    results = []

    # 1. Null values
    null_count = df.isnull().sum().sum()
    if null_count == 0:
        results.append(["Completeness (Null Values)", "PASS", "No null values found"])
    else:
        results.append(["Completeness (Null Values)", "FAIL", f"{null_count} null value(s) found"])

    # 2. Numeric data types
    datatype_errors = []
    for column in numeric_cols:
        converted = pd.to_numeric(df[column], errors="coerce")
        invalid_count = (df[column].notna() & converted.isna()).sum()
        if invalid_count > 0:
            datatype_errors.append(f"{column}: {invalid_count} non-numeric value(s)")
    if len(datatype_errors) == 0:
        results.append(["Data Type Accuracy", "PASS", "All numeric columns contain numbers only"])
    else:
        results.append(["Data Type Accuracy", "FAIL", ", ".join(datatype_errors)])

    # 3. Placeholder text
    special_found = []
    for column in ["Rainfall_mm", "Temperature_C"]:
        for text in ["Unknown/NA", "not_available"]:
            found = (df[column].astype("string").str.contains(text, regex=False, na=False).any())
            if found:
                special_found.append(f"{column} contains {text}")
    if len(special_found) == 0:
        results.append(["Placeholder Text", "PASS", "No placeholder text in numeric columns"])
    else:
        results.append(["Placeholder Text", "FAIL", ", ".join(special_found)])

    # 4. Duplicate Record IDs
    duplicate_count = (df["Record_ID"].value_counts() > 1).sum()
    if duplicate_count == 0:
        results.append(["Uniqueness (Record IDs)", "PASS", "No duplicate Record_ID"])
    else:
        results.append(["Uniqueness (Record IDs)", "FAIL", f"{duplicate_count} duplicate Record_ID value(s)"])

    # 5. Date format
    invalid_dates = []
    for value in df["Observation_Date"].dropna():
        value = str(value).strip()
        if value != "Unknown" and not pd.Series([value]).str.match(date_pattern).iloc[0]:
            invalid_dates.append(value)
    if len(invalid_dates) == 0:
        results.append(["Date Format", "PASS", "All dates are valid or marked Unknown"])
    else:
        results.append(["Date Format", "FAIL", f"{len(invalid_dates)} invalid date(s)"])

    # 6. Categorical values
    categorical_errors = []
    allowed_provinces = {"Benguet", "Bukidnon", "Cagayan", "Davao del Norte", "Iloilo", "Isabela",
                         "Negros Occidental", "Nueva Ecija", "Pangasinan", "Quezon", "Unknown"}
    invalid_provinces = set(df["Province"].dropna().astype(str)) - allowed_provinces
    if len(invalid_provinces) > 0:
        categorical_errors.append(f"Province: {invalid_provinces}")

    allowed_municipalities = {"Ilagan", "Kabankalan", "La Trinidad", "Lucena", "Malaybalay",
                              "Passi", "San Jose", "Tagum", "Tuguegarao", "Urdaneta", "Unknown"}
    invalid_municipalities = set(df["Municipality"].dropna().astype(str)) - allowed_municipalities
    if len(invalid_municipalities) > 0:
        categorical_errors.append(f"Municipality: {invalid_municipalities}")

    allowed_typhoon = {"None", "Low", "Moderate", "High"}
    invalid_typhoon = set(df["Typhoon_Impact"].dropna().astype(str)) - allowed_typhoon
    if len(invalid_typhoon) > 0:
        categorical_errors.append(f"Typhoon_Impact: {invalid_typhoon}")

    if len(categorical_errors) == 0:
        results.append(["Consistency (Categorical Values)", "PASS", "All categorical values are standardized"])
    else:
        results.append(["Consistency (Categorical Values)", "FAIL", ", ".join(categorical_errors)])

    # 7. Valid ranges
    rainfall_numeric = pd.to_numeric(df["Rainfall_mm"], errors="coerce")
    invalid_rainfall = rainfall_numeric[(rainfall_numeric < 0) | (rainfall_numeric > 1000)]
    temperature_numeric = pd.to_numeric(df["Temperature_C"], errors="coerce")
    invalid_temperature = temperature_numeric[(temperature_numeric < 10) | (temperature_numeric > 45)]
    humidity_numeric = pd.to_numeric(df["Relative_Humidity_pct"], errors="coerce")
    invalid_humidity = humidity_numeric[(humidity_numeric < 0) | (humidity_numeric > 100)]
    damage_numeric = pd.to_numeric(df["Crop_Damage_ha"], errors="coerce")
    invalid_damage = damage_numeric[damage_numeric < 0]

    range_errors = []
    if len(invalid_rainfall) > 0:
        range_errors.append(f"Rainfall: {len(invalid_rainfall)}")
    if len(invalid_temperature) > 0:
        range_errors.append(f"Temperature: {len(invalid_temperature)}")
    if len(invalid_humidity) > 0:
        range_errors.append(f"Humidity: {len(invalid_humidity)}")
    if len(invalid_damage) > 0:
        range_errors.append(f"Crop damage: {len(invalid_damage)}")

    if len(range_errors) == 0:
        results.append(["Validity (Numeric Ranges)", "PASS", "All numeric values are within valid ranges"])
    else:
        results.append(["Validity (Numeric Ranges)", "FAIL", "Invalid values - " + ", ".join(range_errors)])

    # 8. Outliers (IQR)
    outlier_details = []
    for column in numeric_cols:
        values = pd.to_numeric(df[column], errors="coerce").dropna()
        q1 = values.quantile(0.25)
        q3 = values.quantile(0.75)
        iqr = q3 - q1
        lower_limit = q1 - 1.5 * iqr
        upper_limit = q3 + 1.5 * iqr
        outliers = values[(values < lower_limit) | (values > upper_limit)]
        if len(outliers) > 0:
            outlier_details.append(f"{column}: {len(outliers)} outlier(s)")
    if len(outlier_details) == 0:
        results.append(["Outlier Review", "PASS", "No outliers detected"])
    else:
        results.append(["Outlier Review", "FAIL", ", ".join(outlier_details)])

    results_df = pd.DataFrame(results, columns=["Check", "Status", "Details"])
    display(results_df)

    if (results_df["Status"] == "PASS").all():
        print("\nOVERALL RESULT: PASS")
        print("All data quality checks passed.")
    else:
        print("\nOVERALL RESULT: FAIL")
        print("Some data quality checks failed.")

    return results_df
```
